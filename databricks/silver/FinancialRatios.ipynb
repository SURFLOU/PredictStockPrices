```python
df = spark.table("plstocks.bronze_financial_ratios")
```

```python
from pyspark.sql.functions import col, isnan, when

# Change NaN to NULL for FLOAT and DOUBLE columns
df = df.select([
    when(isnan(c), None).otherwise(col(c)).alias(c) if dict(df.dtypes)[c] in ['double', 'float'] 
    else col(c).alias(c) 
    for c in df.columns
])

# display(df)
```

```python
df.write.format("delta").mode("overwrite").saveAsTable("plstocks.silver_financial_ratios")
```
