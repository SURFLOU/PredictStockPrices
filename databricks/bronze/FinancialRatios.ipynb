```python
host = dbutils.secrets.get(scope = "postgresql", key = "host")
database = dbutils.secrets.get(scope = "postgresql", key = "database")
user = dbutils.secrets.get(scope = "postgresql", key = "user")
password = dbutils.secrets.get(scope = "postgresql", key = "password")
```

```python
financial_ratios = (spark.read
  .format("postgresql")
  .option("dbtable", "public.financial_ratios")
  .option("host", host)
  .option("port", "5432") 
  .option("database", database)
  .option("user", user) 
  .option("password", password)
  .load()
)
```

```python
display(financial_ratios.limit(15))
```

```python
%python
financial_ratios.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("plstocks.bronze_financial_ratios")
```
